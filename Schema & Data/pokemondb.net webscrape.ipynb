```python
#Imports
from bs4 import BeautifulSoup
import requests,re,pandas
```

```python
#Access site content
response=requests.get('https://pokemondb.net/ability')
soup=BeautifulSoup(response.text,'html.parser')
#Extract Ability Info
info=soup.select('a.ent-name')
```

```python
#Make and add to abilities into list
ability_list=[]
for i in info:
    line=str(i)
    content=re.search(">[a-zA-Z0-9_\s'-]*<",line).group(0)
    ability_list.append(content[1:-1])
```

```python
#Create dataset
ability_csv=[]
counter=1
for i in ability_list:
    data={'AbilityID':counter,'AbilityName':i}
    ability_csv.append(data)
    counter+=1
#make dataframe
df=pandas.DataFrame(ability_csv)
df.to_csv('Ability.csv',index=False)
```

```python
#Put into text file
with open('ability.txt','w') as file:
    for i in ability_list:
        file.write(f'Name:{i}\n')
```

```python
#Put into MySql file
counter=0
endline=','
with open('Ability_Data.sql','w') as file:
    file.write('''USE pokemon_storage;
INSERT INTO ability
(AbilityID,AbilityName)
VALUES''')
    for i in ability_list:
        if counter==len(ability_list)-1:
            endline=';'
        file.write(f'''({counter},"{i}"){endline}\n''')
        counter+=1
```

```python
#Access site content
response2=requests.get('https://pokemondb.net/item/all')
soup2=BeautifulSoup(response2.text,'html.parser')
#Extract Ability Info
info2=soup2.select('a.ent-name')
```

```python
#Make and add to item list
item_list=[]
for i in info2:
    line2=str(i)
    content2=str(re.search(">[a-zA-Z0-9_\s'-éñ]*<",line2))
    if content2!='None':
        item_list.append(content2[41:-3])
```

```python
#Make dataset
item_csv=[]
item_csv.append({'ItemID':0,'ItemName':'None'})
counter=1
for i in item_list:
    data={'ItemID':counter,'ItemName':i}
    item_csv.append(data)
    counter+=1
#make dataframe
df2=pandas.DataFrame(item_csv)
df2.to_csv('Item.csv',index=False)
```

```python
#Put into text file
with open('item.txt','w',encoding='utf8') as file:
    for i in item_list:
        file.write(f'Name:{i}\n')
```

```python
#Put into MySql file
counter=1
endline=','
with open('Item_Data.sql','w',encoding='utf8') as file:
    file.write('''USE pokemon_storage;
INSERT INTO items
(ItemID,ItemName)
VALUES
(0,"None"),\n''')
    for i in item_list:
        if counter==len(item_list):
            endline=';'
        file.write(f'''({counter},"{i}"){endline}\n''')
        counter+=1
```

```python
#Access site content
response3=requests.get('https://pokemondb.net/pokedex/national')
soup3=BeautifulSoup(response3.text,'html.parser')
#Extract pokemon info
info3=soup3.select('span.infocard-lg-data.text-muted')
```

```python
#Make and add to name list
name_list=[]
for i in info3:
    name_list.append(i.get_text()[6:].split())    
```

```python
#Cleanup
name_list_cleaned=[]
for i in name_list:
    if len(i)==2:
        name_list_cleaned.append(i)
    elif len(i)==3:
        name=i[0]+' '+i[1]
        name_list_cleaned.append([name,i[2]])
    elif len(i)==4:
        name_list_cleaned.append([i[0],i[1],i[3]])
    elif len(i)==5:
        name=i[0]+' '+i[1]
        name_list_cleaned.append([name,i[2],i[4]])
```

```python
#Make dataset
name_csv=[]
counter=1
for i in name_list_cleaned:
    if len(i)==2:
        data={'PokeID':counter,'PokedexNo':counter,'PokeName':i[0],'Type1':i[1],'Type2':None}
    elif len(i)==3:
        data={'PokeID':counter,'PokedexNo':counter,'PokeName':i[0],'Type1':i[1],'Type2':i[2]}
    name_csv.append(data)
    counter+=1
#Make datframe
df3=pandas.DataFrame(name_csv)
df3.to_csv('Pokeinfo.csv',index=False)
```

```python
#Put to text file
with open('pokeinfo.txt','w',encoding='utf8') as file:
    for i in name_list_cleaned:
        if len(i)==2:
            file.write(f'Name:{i[0]}, Type:{i[1]}\n')
        elif len(i)==3:
            file.write(f'Name:{i[0]}, Type:{i[1]}/{i[2]}\n')
```

```python
#Put into MySQL file
counter=1
endline=','
with open('Poke_Table.sql','w',encoding='utf8') as file:
    file.write('''USE Pokemon_Storage;
INSERT INTO Pokemon
(PokeID,PokedexNo,PokeName,Type1,Type2)
Values''')
    for i in name_list_cleaned:
        if counter==len(name_list_cleaned):
            endline=';'
        if len(i)==2:
            file.write(f'''({counter},{counter},"{i[0]}","{i[1]}",NULL){endline}\n''')
        elif len(i)==3:
            file.write(f'''({counter},{counter},"{i[0]}","{i[1]}","{i[2]}"){endline}\n''')
        counter+=1
```

```python
#Access site content
response4=requests.get('https://pokemondb.net/mechanics/natures')
soup4=BeautifulSoup(response4.text,'html.parser')
#Extract pokemon info
info4=soup4.select('td.ent-name')
```

```python
#Make and add nature list
nature_list=[]
for i in info4:
    line4=str(i)
    content4=re.search(">[a-zA-Z0-9_\s'-]*<",line4).group(0)
    nature_list.append(content4[1:-2])
```

```python
#Make dataset
nature_csv=[]
counter=1
for i in nature_list:
    data={'NatureID':counter,'NatureName':i}
    nature_csv.append(data)
    counter+=1
#Make dataframe
df4=pandas.DataFrame(nature_csv)
df4.to_csv('Nature.csv',index=False)
```

```python
#Put to text file
with open('nature.txt','w') as file:
    for i in nature_list:
        file.write(f'{i}\n')
```

```python
#Put into MySQL file
counter=0
endline=','
with open('Poke_Nature_Data.sql','w') as file:
    file.write('''USE Pokemon_Storage;
INSERT INTO Nature
(NatureID,NatureName)
Values''')
    for i in nature_list:
        if counter==len(nature_list)-1:
            endline=';'
        file.write(f'''({counter},"{i}"){endline}\n''')
        counter+=1
```
